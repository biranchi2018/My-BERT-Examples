### BERT for Sequence Classification

```
from transformers import BertTokenizer, BertForSequenceClassification
import torch
```

```
# Load pre-trained BERT tokenizer and model
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
model = BertForSequenceClassification.from_pretrained("bert-base-uncased")
```

```
# Example input sentence
text = "Hugging Face is creating amazing tools for NLP!"

# Tokenize the input text
inputs = tokenizer(text, return_tensors="pt")
inputs
```

```
# Perform classification (forward pass)
with torch.no_grad():
    outputs = model(**inputs)

outputs
```

```
# The output contains logits, which are raw scores for each class
logits = outputs.logits
logits
```

```
# Convert logits to predicted class
predicted_class = torch.argmax(logits, dim=-1)
print(f"Predicted class: {predicted_class.item()}")
```

For BertForSequenceClassification, the labels are typically task-specific. If the model was trained on a binary classification task (e.g., sentiment analysis), then:

The first logit (0.4061) would usually correspond to class 0 (e.g., negative or neutral sentiment).
The second logit (0.0188) would usually correspond to class 1 (e.g., positive sentiment).


```
label_mapping = model.config.id2label
print("Label Mapping:")
print(label_mapping)
```

```
# Map the predicted class ID to a human-readable label
predicted_label = label_mapping[predicted_class.item()]
print(f"Predicted label: {predicted_label}")
```
